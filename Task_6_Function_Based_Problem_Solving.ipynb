```python
# Problem Statement 1:Smart Parking Lot Management System

def check_parking(max_spots, parking_status):
    total_now = 0

    for action in parking_status:
        if action == "IN":
            total_now = total_now + 1
        elif action == "OUT":
            total_now = total_now - 1

    # Figure out if we have room left
    if total_now >= max_spots:
        my_status = "Full / Over Capacity"
    else:
        my_status = "Available"

    print(f"Currently Parked Vehicles: {total_now}")
    print(f"Parking Status: {my_status}")

# Testing
check_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
# Problem Statement 2:Online Food Delivery Time Estimator

def estimate_delivery_time(distance_km, traffic_level, weather_condition):
    estimated_time = distance_km * 3

    traffic_level = traffic_level.lower()
    if traffic_level == "none":
        pass
    elif traffic_level == "low":
        estimated_time += 5  # Add 5 minutes for low traffic
    elif traffic_level == "medium":
        estimated_time += 10 # Add 10 minutes for medium traffic
    elif traffic_level == "high":
        estimated_time += 15 # Add 15 minutes for high traffic
    else:
        print(f"Warning: Unknown traffic level '{traffic_level}'. No traffic delay applied.")

    weather_condition = weather_condition.lower()
    if weather_condition == "clear":
        pass
    elif weather_condition == "rainy":
        estimated_time += 16 # Add 16 minutes for rainy weather
    elif weather_condition == "snowy":
        estimated_time += 25 # Add 25 minutes for snowy weather
    elif weather_condition == "stormy":
        estimated_time += 40 # Add 40 minutes for stormy weather
    else:
        print(f"Warning: Unknown weather condition '{weather_condition}'. No weather delay applied.")

    return estimated_time

# Sample Input from problem statement
sample_distance = 8
sample_traffic = "High"
sample_weather = "Rainy"

# Calculate and display the estimated delivery time
final_eta = estimate_delivery_time(sample_distance, sample_traffic, sample_weather)
print(f"Estimated Delivery Time: {final_eta} minutes")
```

```python
# Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

def analyze_seat_occupancy(total_seats, booked_seats):
    num_booked_seats = len(booked_seats)
    occupancy_percentage = (num_booked_seats / total_seats) * 100

    show_status = ""
    additional_show_suggestion = ""

    if occupancy_percentage == 100:
        show_status = "Housefull"
        additional_show_suggestion = "Consider opening additional shows!"
    elif occupancy_percentage >= 75:
        show_status = "Almost Full"
    else:
        show_status = "Available"

    print(f"Occupancy: {occupancy_percentage:.0f}%")
    print(f"Show Status: {show_status}")
    if additional_show_suggestion:
        print(additional_show_suggestion)

total_seats_sample = 200
booked_seats_sample = [1] * 150

analyze_seat_occupancy(total_seats_sample, booked_seats_sample)
```

```python
# Problem Statement 4:Cloud Server Load Classification System

def classify_server_load(cpu_readings):

    # Calculate average
    average_cpu_load = sum(cpu_readings) / len(cpu_readings)

    server_status = ""
    if average_cpu_load < 50:
        server_status = "Normal"
    elif 50 <= average_cpu_load < 80:
        server_status = "Warning"
    else:
        server_status = "Critical"

    print(f"Average CPU Load: {average_cpu_load:.0f}%")
    print(f"Server Status: {server_status}")

# Sample Input
sample_cpu_readings = [45, 60, 70, 85, 90]

# Test the function with the sample input
classify_server_load(sample_cpu_readings)
```

```python
# Problem Statement 5:Smart Classroom Resource Usage Monitor

def monitor_classroom_resources(resource_usage):
    overused_resources = []
    energy_alert = "No"

    # Define a threshold for overuse (e.g., more than 8 hours for any resource)
    overuse_threshold = 8

    for resource, hours in resource_usage.items():
        if hours > overuse_threshold:
            overused_resources.append(resource)
            energy_alert = "Yes"

    # Output results
    if overused_resources:
        print(f"Overused Resources: {', '.join(overused_resources)}")
    else:
        print("Overused Resources: None")
    print(f"Energy Alert: {energy_alert}")

# Sample Input
sample_resource_usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_classroom_resources(sample_resource_usage)
```

```python
# Problem Statement 6: Online Event Registration Capacity Controller

def manage_event_registrations(event_capacity, total_registrations):
    confirmed_registrations = min(event_capacity, total_registrations)
    waitlisted_users = max(0, total_registrations - event_capacity)

    if total_registrations >= event_capacity:
        registration_status = "Closed"
    else:
        registration_status = "Open"

    print(f"Confirmed Registrations: {confirmed_registrations}")
    print(f"Waitlisted Users: {waitlisted_users}")
    print(f"Registration Status: {registration_status}")

# Sample Input
sample_event_capacity = 100
sample_registrations = 105

manage_event_registrations(sample_event_capacity, sample_registrations)
```
