```python
# Problem 1: Smart Light Controller

class smart_light:
  def __init__(self, light_name, action):
    self.light_name = light_name
    self.action = action

  def get_action_status(self):
    if self.action == "on":
      return f"{self.light_name} is ON"
    elif self.action == "off":
      return f"{self.light_name} is OFF"
    else:
      return "Invalid action"

lightname =  "Bedroom Light"
action = "on"

smart_light_instance = smart_light(lightname, action)
print(smart_light_instance.get_action_status())
```

```python
# Problem 2: Employee ID Card System

class employee:
  def __init__(self, employee_name, employee_id, department):
    self.employee_name = employee_name
    self.employee_id = employee_id
    self.department = department

  def get_employee_details(self):
    return f"Employee ID Card\nName: {self.employee_name}\nID: {self.employee_id}\nDepartment: {self.department}"


employee_name = "Rahul"
employee_id = "EMP102"
department = "IT"


# Create an instance of the employee class
employee_instance = employee(employee_name, employee_id, department)
# Call the method on the instance
print(employee_instance.get_employee_details())
```

```python
# Problem 3: Mobile Contact Record

class Contact:
  def __init__(self, contact_name, phone_number):
    self.contact_name = contact_name
    self.phone_number = phone_number

  def display_contact_info(self):
    return f"Contact Saved\nName: {self.contact_name}\nPhone: {self.phone_number}"


contact_name = "Anita"
phone_number = "9876543210"

contact_instance = Contact(contact_name, phone_number)

print(contact_instance.display_contact_info())
```

```python
# Problem 4: Product Price Tag Generator

class ProductPriceTag:
  def __init__(self, product_name, price):
    self.product_name = product_name
    self.price = price

  def generate_price_tag(self):
    return f"Product: {self.product_name}\nPrice: ₹{self.price}"

product_name = "Headphones"
price = 2499

price_tag_instance = ProductPriceTag(product_name, price)

print(price_tag_instance.generate_price_tag())
```

```python
# Problem 5: Movie Rating Display System

class MovieRating:
  def __init__(self, movie_name, rating):
    self.movie_name = movie_name
    self.rating = rating

  def display_movie_details(self):
    return f"Movie: {self.movie_name}\nRating: {self.rating} / 5"


# Sample Input
movie_name = "Inception"
rating = 4.8
movie_instance = MovieRating(movie_name, rating)
print(movie_instance.display_movie_details())
```

```python
# Problem 6: Delivery Address Manager

class Customer_details:
  def __init__(self, name, address):
    self.name = name
    self.address = address

  def display_address(self):
    return f"Delivery Details\nCustomer: {self.name}\nAddress: {self.address}"

name = "Suman"
address = "Hyderabad"

customer_instance = Customer_details(name, address)

print(customer_instance.display_address())
```
