```python
# Problem 1: Social Media – Post Engagement Analyzer
def analyze_post_engagement():
    n = int(input("Enter number of like entries: "))

    total_likes = 0

    for i in range(n):
        likes = int(input("Enter likes: "))
        total_likes += likes

    print("Total Likes:", total_likes)

    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")


analyze_post_engagement()

```

```python
# Problem 2: Healthcare – Medicine Stock Alert System

def medicine_stock_alert(stock):
  if stock < 10:
    return "Low Stock Alert"
  elif stock >= 10:
    return "Stock Sufficient"
  else:
    return "Invalid Stock"

stock = int(input("Enter stock: "))
print()

print(f"Medicine Stock: {stock}")
print(f"status: {medicine_stock_alert(stock)}")
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker

def rainfall_adequacy_checker(rainfall):
    average_rainfall = sum(rainfall) / len(rainfall)
    required_level = 50

    print(f"Average Rainfall: {average_rainfall}")

    if average_rainfall >= required_level:
        return "Adequate Rainfall"
    else:
        return "Inadequate Rainfall"


rainfall = []

days = int(input("Enter number of days: "))

for i in range(days):
    n = input(f"Enter rainfall for day {i+1}: ")
    rainfall.append(int(n))

result = (f"Rainfall Status: {rainfall_adequacy_checker(rainfall)}")
print(result)



```

```python
# Problem 4: Social Media – Duplicate Account Detection

def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)

    if len(unique_users) != len(usernames):
        return "Duplicate Accounts Found: Yes"
    else:
        return "Duplicate Accounts Found: No"


# Predefined usernames list
usernames = ["john", "emma", "alex", "john", "mike"]

result = detect_duplicate_accounts(usernames)
print(result)
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter

def premium_crop_filter(prices):
  premium_crops = []
  for price in prices:
    if price > 2000:
      premium_crops.append(price)
  return premium_crops


prices = [1000, 2500, 3200, 2000, 650]

print(f"Premium Crops: {premium_crop_filter(prices)}")
```

```python
# Problem 7: System Monitoring – Application Health Checker

def application_health_checker(errors):
    if errors == 0:
        return "Healthy"
    elif 0 < errors <= 5:
        return "Minor Issues"
    else:
        return "Critical Issues"

# Example usage:
error_count = 7
status = application_health_checker(error_count)
print(f"Error Count: {error_count}")
print(f"System Status: {status}")
```

```python
# Problem 8: Banking – Daily Transaction Limit Checker

def daily_transaction_limit_checker(amount):
    daily_limit = 50000
    if amount <= daily_limit:
        return "Approved"
    else:
        return "Rejected"

# Example usage:
transaction_amount = 60000
status = daily_transaction_limit_checker(transaction_amount)
print(f"Transaction Amount: {transaction_amount}")
print(f"Transaction Status: {status}")
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System

def check_attendance_eligibility(attendance_records):
    total_days = len(attendance_records)
    if total_days == 0:
        return 0.0, "Not Applicable"

    present_days = sum(attendance_records)
    attendance_percentage = (present_days / total_days) * 100

    if attendance_percentage >= 75:
        eligibility_status = "Eligible"
    else:
        eligibility_status = "Not Eligible"

    return attendance_percentage, eligibility_status

# Example usage:
student_attendance = [1, 1, 1, 0, 1, 1, 1, 1, 0, 1]

percentage, eligibility = check_attendance_eligibility(student_attendance)

print(f"Attendance Percentage: {percentage:.1f}")
print(f"Exam Eligibility: {eligibility}")
```

```python
# Problem 10: Smart Electricity Bill Analyzer

def calculate_electricity_bill_and_usage(units):
    total_bill = 0

    # Calculate bill based on units consumed
    if units <= 100:
        total_bill = units * 3
    elif units <= 200:
        total_bill = (100 * 3) + ((units - 100) * 5)
    else:
        total_bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Classify usage level
    usage_status = ""
    if total_bill < 500:
        usage_status = "Low Usage"
    elif 500 <= total_bill <= 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"

    return total_bill, usage_status

# Example usage:
units_consumed = 250
bill, usage = calculate_electricity_bill_and_usage(units_consumed)

print(f"Units Consumed: {units_consumed}")
print(f"Total Bill: ₹{bill}")
print(f"Usage Status: {usage}")
```
