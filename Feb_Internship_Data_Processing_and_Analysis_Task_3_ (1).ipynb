# **Problem Statement 1: Employee Performance Bonus Eligibility**
**Description:**

A company evaluates employee performance scores at the end of the year. You are given a dictionary containing employee names and their performance scores.

Requirements:

Identify the highest performance score.

Handle ties if multiple employees have the same highest score.

Display all employees eligible for the top performance bonus.

Input:

employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}

Expected Output:

Top Performers Eligible for Bonus: Ravi, Kiran (Score: 92)




```python
# Define a dictionary of employees and their performance scores
employees = { "Ravi": 92, "Anita": 88, "Kiran": 92, "Suresh": 85 }
# Find the highest performance score among all employees
highest_score = max(employees.values())
# Identify all employees who achieved the highest score (handling ties)
top_score_employess = [names for names in employees if employees[names] == highest_score]
# Print the names of the top performers and their score
print(f"Top Performers Eligible for Bonus: {', '.join(top_score_employess)} (Score: {highest_score})")
```

# **Problem Statement 2: Search Query Keyword Analysis**

**Description:**

An e-commerce website stores customer search queries. You are given a search query sentence entered by a user.

Requirements:

Convert the input to lowercase.

Ignore common punctuation.

Count the frequency of each keyword.

Display only keywords searched more than once.

Input:

"Buy mobile phone buy phone online"

Expected Output:

{'buy': 2, 'phone': 2}


```python
# Define the input search query string
string = "Buy mobile phone buy phone online"
# Convert the entire string to lowercase for case-insensitive analysis
n_string = string.lower()
# Split the string into individual words (keywords)
string = n_string.split()
# Initialize an empty dictionary to store keyword frequencies
freq = {}
# Iterate through each word in the list of keywords
for i in string:
  # Check if the current word appears more than once
  if string.count(i) > 1:
    # If it does, store its frequency in the dictionary
    freq[i] = string.count(i)
# Print the dictionary containing keywords and their frequencies (only for those appearing more than once)
print(freq)
```

# **Problem Statement 3: Sensor Data Validation**
Description:

A factory collects sensor readings every hour. Each reading is stored in a list where the index represents the hour and the value represents the sensor reading.

Requirements:

Identify readings that are even numbers (valid readings).

Store them as (hour_index, reading_value) pairs.

Ignore odd readings (invalid readings).

Input:

sensor_readings = [3, 4, 7, 8, 10, 12, 5]

Expected Output:

Valid Sensor Readings (Hour, Value):

[(1, 4), (3, 8), (4, 10), (5, 12)]





```python
# Define a list of sensor readings
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
# Initialize an empty list to store valid sensor readings (hour_index, reading_value)
valid_data = []
# Iterate through the sensor readings with their indices
for i in range(len(sensor_readings)):
  # Check if the current sensor reading is an even number
  if sensor_readings[i] %2==0:
    # If it's even, append a tuple of (hour_index, reading_value) to valid_data
    valid_data.append((i, sensor_readings[i]))
# Print the list of valid sensor readings
print(valid_data)
```

# **Problem Statement 4: Email Domain Usage Analysis**

Description:

A company wants to analyze which email providers its users are using. You are given a list of employee email IDs.

Requirements:

Count how many users belong to each email domain.

Calculate the percentage usage of each domain.

Input:

emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]

Expected Output:

gmail.com: 60%

yahoo.com: 40%


```python
# Define a list of employee email IDs
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]

# Extract the domain from each email address using a list comprehension
domains_list = [email.split("@")[1] for email in emails]
# Iterate through unique domains to calculate their usage percentage
for i in set(domains_list):
  # Calculate the percentage of users for the current domain
  percentage = (domains_list.count(i)/len(domains_list))*100
  # Print the domain and its usage percentage, formatted as an integer
  print(f"{i}: {int(percentage)}%")
```

# **Problem Statement 5: Sales Spike Detection**

Description:

A retail company tracks daily sales. Sudden spikes in sales may indicate promotions or unusual activity.

Requirements:

Calculate the average daily sales.

Detect days where sales are more than 30% above average.

Display the day number and sale value.

Input:

sales = [1200, 1500, 900, 2200, 1400, 3000]

Expected Output:

Day 4: 2200

Day 6: 3000


```python
# Define a list of daily sales figures
sales = [1200, 1500, 900, 2200, 1400, 3000]
# Calculate the average daily sales
average = sum(sales)/len(sales)
# Iterate through the sales data with day indices
for i in range(len(sales)):
  # Check if the current day's sales are above the average
  if sales[i] > average:
    # If sales are above average, print the day number (1-indexed) and sales value
    print(f"Day {i+1}: {sales[i]} ")
```

# **Problem Statement 6: Duplicate User ID Detection**

Description:

A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.

Requirements:

Identify duplicate user IDs.

Display how many times each duplicate appears.

Input:

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

Expected Output:

user1 → 3 times

user3 → 2 times



```python
# Define a list of user IDs
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Initialize an empty list to keep track of user IDs that have already been processed and printed
printed = []

# Iterate through each user ID in the list
for user in user_ids:
    # Check if the current user ID appears more than once in the list
    # and if it has not been printed already to avoid redundant output
    if user_ids.count(user) > 1 and user not in printed:
        # Print the user ID and the number of times it appears
        print(f"{user} → {user_ids.count(user)} times")
        # Add the user ID to the 'printed' list to ensure it's only reported once
        printed.append(user)
```
