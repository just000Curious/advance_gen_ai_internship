# **Pass / Fail Analyzer**
Problem Statement:
Analyze student results.

marks = [45, 78, 90, 33, 60]
Requirements:

A student passes if marks ≥ 50

Count the total number of pass students

Count the total number of fail students

Print the final result clearly



```python
marks = [45, 78, 90, 33, 60]
def pass_fail(marks):
  pass_students = 0
  fail_student = 0
  for mark in marks:
    if mark >= 50:
      pass_students += 1
    else:
      fail_student += 1

  print("results = ", pass_students, "pass students")
  print("results = ", fail_student, "fail students")
pass_fail(marks)


```

# **Simple Data Cleaner**
Problem Statement:
Clean and standardize user names.

names = [" Alice ", "bob", " CHARLIE "]

Requirements:

Remove extra spaces

Convert all names to lowercase

Print the cleaned list


```python
names = [" Alice ", "bob", " CHARLIE "]
def clean_data(names):
  for i in range(len(names)):
    names[i] = names[i].strip()
    names[i] = names[i].lower()
  print(names)
clean_data(names)
```

# **Error Message Detector**
Problem Statement:

Detect error messages from system logs.

logs = ["INFO", "ERROR", "WARNING", "ERROR"]

Requirements:

  Count the number of "ERROR" entries

  Print the total error count


```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

def error_message(log):
  error_count = 0
  for log in logs:
    if log == "ERROR":
      error_count += 1
  print(f"the total error count is {error_count}")
error_message(logs)

```
